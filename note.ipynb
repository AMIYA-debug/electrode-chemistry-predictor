```python
import pandas as pd
df = pd.read_csv("weld_dataset.csv")
df
```

```python
df = df.drop("Sample", axis=1)
df.head()
```

```python
x = df.drop(["C","Si","Mn","Cr","Ni","Mo","Ti","Al","Nb","S","P"],axis=1)
y0= df[["C"]]
y1 = df[["Si"]]
y2 = df[["Mn"]]
y3 = df[["Cr"]]
y4 = df[["Ni"]]
y5 = df[["Mo"]]
y6 = df[["Ti"]]
y7 = df[['Al']]
y8 = df[["Nb"]]
y9 = df[["S"]]
y10 = df[["P"]]
```

```python
from sklearn.model_selection import train_test_split
x0_train , x0_test , y0_train , y0_test = train_test_split(x,y0, test_size=0.2, random_state=42)
x1_train , x1_test , y1_train , y1_test = train_test_split(x,y1, test_size=0.2, random_state=42)
x2_train , x2_test , y2_train , y2_test = train_test_split(x,y2, test_size=0.2, random_state=42)
x3_train , x3_test , y3_train , y3_test = train_test_split(x,y3, test_size=0.2, random_state=42)
x4_train , x4_test , y4_train , y4_test = train_test_split(x,y4, test_size=0.2, random_state=42)
x5_train , x5_test , y5_train , y5_test = train_test_split(x,y5, test_size=0.2, random_state=42)
x6_train , x6_test , y6_train , y6_test = train_test_split(x,y6, test_size=0.2, random_state=42)
x7_train , x7_test , y7_train , y7_test = train_test_split(x,y7, test_size=0.2, random_state=42)
x8_train , x8_test , y8_train , y8_test = train_test_split(x,y8, test_size=0.2, random_state=42)
x9_train , x9_test , y9_train , y9_test = train_test_split(x,y9, test_size=0.2, random_state=42)
x10_train , x10_test , y10_train , y10_test = train_test_split(x,y10, test_size=0.2, random_state=42)
```

```python
from sklearn.base import clone
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import ConstantKernel, Matern, RBF, WhiteKernel
from sklearn.linear_model import ElasticNet, Lasso, LinearRegression, Ridge
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
import numpy as np
```

```python
target_datasets = {
    "C":  (x0_train ,x0_test,y0_train,y0_test),
    "Si": (x1_train, x1_test, y1_train, y1_test),
    "Mn": (x2_train, x2_test, y2_train, y2_test),
    "Cr": (x3_train, x3_test, y3_train, y3_test),
    "Ni": (x4_train, x4_test, y4_train, y4_test),
    "Mo": (x5_train, x5_test, y5_train, y5_test),
    "Ti": (x6_train, x6_test, y6_train, y6_test),
    "Al": (x7_train, x7_test, y7_train, y7_test),
    "Nb": (x8_train, x8_test, y8_train, y8_test),
    "S": (x9_train, x9_test, y9_train, y9_test),
    "P": (x10_train, x10_test, y10_train, y10_test),
}
```

```python
candidate_models = {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=0.01, max_iter=10_000),
    "ElasticNet": ElasticNet(alpha=0.01, l1_ratio=0.5, max_iter=10_000),
    "Gaussian Process (RBF)": GaussianProcessRegressor(
        kernel=ConstantKernel(1.0) * RBF(length_scale=1.0) + WhiteKernel(noise_level=1e-5),
        alpha=1e-6,
        normalize_y=True,
        n_restarts_optimizer=2,
        random_state=42,
    ),
    "Gaussian Process (Matern)": GaussianProcessRegressor(
        kernel=ConstantKernel(1.0) * Matern(length_scale=1.0, nu=1.5) + WhiteKernel(noise_level=1e-5),
        alpha=1e-6,
        normalize_y=True,
        n_restarts_optimizer=2,
        random_state=42,
    ),
    "SVR (RBF)": SVR(kernel="rbf", C=1.0, epsilon=0.05),
    "KNN": KNeighborsRegressor(n_neighbors=3, weights="distance"),
    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        min_samples_leaf=2,
        random_state=42,
    ),
    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=4,
        min_samples_leaf=1,
        random_state=42,
        n_jobs=-1,
    ),
    "Extra Trees": ExtraTreesRegressor(
        n_estimators=200,
        max_depth=4,
        min_samples_leaf=1,
        random_state=42,
        n_jobs=-1,
    ),
}
```

```python
feature_scalers = {}
fitted_models = {}
test_predictions = {}
evaluation_rows = []
```

```python
for target_name, (x_train, x_test, y_train, y_test) in target_datasets.items():
    
    scaler = StandardScaler()
    x_train_scaled = scaler.fit_transform(x_train)
    x_test_scaled = scaler.transform(x_test)

    feature_scalers[target_name] = scaler
    fitted_models[target_name] = {}
    test_predictions[target_name] = {}

    
    y_train_array = np.asarray(y_train).ravel()
    y_test_array = np.asarray(y_test).ravel()

    for model_name, model in candidate_models.items():
        current_model = clone(model)
        current_model.fit(x_train_scaled, y_train_array)

        predictions = current_model.predict(x_test_scaled)

        fitted_models[target_name][model_name] = current_model
        test_predictions[target_name][model_name] = predictions

        evaluation_rows.append(
            {
                "Target": target_name,
                "Model": model_name,
                "RMSE": root_mean_squared_error(y_test_array, predictions),
                "MAE": mean_absolute_error(y_test_array, predictions),
                "R2": r2_score(y_test_array, predictions),
            }
        )
```

```python
model_results = (
    pd.DataFrame(evaluation_rows)
    .sort_values(
        by=["Target", "RMSE", "MAE", "R2"],
        ascending=[True, True, True, False],
    )
    .reset_index(drop=True)
)

best_models_by_target = (
    model_results
    .groupby("Target", as_index=False)
    .first()
    .sort_values("Target")
    .reset_index(drop=True)
)

print("All model results:")
display(model_results)

print("Best model for each target, selected primarily by lowest RMSE:")
display(best_models_by_target)

```

```python
from pathlib import Path
import json

import joblib
from sklearn.pipeline import Pipeline

models_dir = Path("models")
models_dir.mkdir(exist_ok=True)


feature_names = list(x1_train.columns)

for target_name, (x_train, _, _, _) in target_datasets.items():
    if list(x_train.columns) != feature_names:
        raise ValueError(
            f"Feature order for {target_name} differs from the saved training order."
        )

if len(feature_names) != len(set(feature_names)):
    raise ValueError("Training feature names contain duplicates.")

(models_dir / "feature_names.json").write_text(
    json.dumps(feature_names, indent=2),
    encoding="utf-8",
)

export_manifest = {}

for selected in best_models_by_target.itertuples(index=False):
    target_name = str(selected.Target)
    model_name = str(selected.Model)

    if target_name not in fitted_models:
        raise KeyError(f"No fitted models found for target '{target_name}'.")
    if model_name not in fitted_models[target_name]:
        raise KeyError(
            f"Best model '{model_name}' for '{target_name}' is not present "
            "in fitted_models."
        )

    best_model = fitted_models[target_name][model_name]
    target_dir = models_dir / target_name
    target_dir.mkdir(exist_ok=True)

    model_path = target_dir / "model.pkl"
    scaler_path = target_dir / "scaler.pkl"


    joblib.dump(best_model, model_path)


    if isinstance(best_model, Pipeline):
        if scaler_path.exists():
            scaler_path.unlink()
        preprocessing = "pipeline"
    else:
        if target_name not in feature_scalers:
            raise KeyError(f"No fitted StandardScaler found for target '{target_name}'.")
        joblib.dump(feature_scalers[target_name], scaler_path)
        preprocessing = "separate_standard_scaler"

    export_manifest[target_name] = {
        "best_model": model_name,
        "model_file": str(model_path),
        "preprocessing": preprocessing,
    }

(models_dir / "best_models.json").write_text(
    json.dumps(export_manifest, indent=2),
    encoding="utf-8",
)

print(f"Exported {len(export_manifest)} best models to: {models_dir.resolve()}")
```
